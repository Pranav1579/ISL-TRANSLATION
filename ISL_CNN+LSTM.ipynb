```python
import os
import shutil
from glob import glob
from collections import defaultdict
from tqdm import tqdm

from PIL import Image

def normalize_class_name(name):
    """Remove suffixes like __copy, __dup, etc."""
    parts = name.split("__")[0]
    return parts.strip()

def gather_class_folders(dataset_dir):
    """Group folders by base class name"""
    class_groups = defaultdict(list)
    for folder in os.listdir(dataset_dir):
        full_path = os.path.join(dataset_dir, folder)
        if os.path.isdir(full_path):
            base_name = normalize_class_name(folder)
            class_groups[base_name].append(full_path)
    return class_groups

def merge_and_redistribute(class_groups, output_dir):
    if os.path.exists(output_dir):
        shutil.rmtree(output_dir)
    os.makedirs(output_dir, exist_ok=True)

    for class_name, folders in tqdm(class_groups.items(), desc="Processing classes"):
        all_frames = []
        for folder in sorted(folders):
            images = sorted(glob(os.path.join(folder, "*.jpg")))
            all_frames.extend(images)

        if len(all_frames) < 128:
            print(f"⚠️ Skipping {class_name}: only {len(all_frames)} frames found.")
            continue

        # Use only first 128 frames
        all_frames = all_frames[:128]

        for i in range(4):
            new_folder = os.path.join(output_dir, f"{class_name}__vid{i}")
            os.makedirs(new_folder, exist_ok=True)
            for j in range(32):
                src = all_frames[i * 32 + j]
                dst = os.path.join(new_folder, f"{j+1:05d}.jpg")
                shutil.copy(src, dst)

    print(f"\n✅ Done! Cleaned dataset saved to: {output_dir}")

# 🏁 Run
dataset_dir = "/content/Cleaned_SignLanguageDataset"
output_dir = "/content/Cleaned_MergedDataset"

class_groups = gather_class_folders(dataset_dir)
merge_and_redistribute(class_groups, output_dir)
```

```python
import os
import shutil
import re
from tqdm import tqdm

def normalize_class_name(name):
    """
    Removes suffixes like '__copy', '___dup', '_vid', '_2', etc.
    """
    name = name.lower().strip()
    name = re.sub(r'(__copy|___dup|_vid|_copy|\d+)+$', '', name)  # remove trailing suffixes
    name = re.sub(r'[_\W]+$', '', name)  # cleanup any remaining symbols
    return name.strip()

def group_and_rename_classes(src_root, dest_root):
    os.makedirs(dest_root, exist_ok=True)
    class_map = {}

    # Scan and normalize class folders
    for folder in tqdm(os.listdir(src_root), desc="Scanning classes"):
        folder_path = os.path.join(src_root, folder)
        if not os.path.isdir(folder_path):
            continue
        clean_name = normalize_class_name(folder)
        class_map.setdefault(clean_name, []).append(folder_path)

    # Copy and group under normalized names
    for class_name, video_folders in tqdm(class_map.items(), desc="Grouping videos"):
        class_dest = os.path.join(dest_root, class_name)
        os.makedirs(class_dest, exist_ok=True)

        for idx, video_path in enumerate(video_folders, start=1):
            dest_path = os.path.join(class_dest, f"video_{idx}")
            shutil.copytree(video_path, dest_path)

    print(f"\n✅ Cleaned and grouped {len(class_map)} classes at: {dest_root}")

# 🔁 Run this
src_dataset = "/content/Cleaned_MergedDataset"
dest_dataset = "/content/GroupedDataset_Renamed"
group_and_rename_classes(src_dataset, dest_dataset)
```

```python
# Install dependencies
# !pip install torch torchvision numpy pillow tqdm

import os
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision.transforms as transforms
import torchvision.models as models
from torch.utils.data import Dataset, DataLoader
from PIL import Image
import numpy as np
from tqdm import tqdm
from collections import Counter

# Set device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# Enhanced Dataset Class
class SignLanguageDataset(Dataset):
    def __init__(self, root_dir, transform=None, seq_length=32):
        self.root_dir = root_dir
        self.transform = transform
        self.seq_length = seq_length
        self.classes = sorted(os.listdir(root_dir))
        self.label_map = {cls: idx for idx, cls in enumerate(self.classes)}

        self.data = []
        for cls in self.classes:
            cls_path = os.path.join(root_dir, cls)
            frames = sorted([f for f in os.listdir(cls_path) if f.endswith(".jpg")],
                            key=lambda x: int(x.split('_')[-1].split('.')[0]))
            self.data.extend([(os.path.join(cls_path, f), self.label_map[cls]) for f in frames])

        # Calculate class weights
        labels = [item[1] for item in self.data]
        class_counts = np.bincount(labels)
        self.class_weights = 1. / torch.tensor(class_counts, dtype=torch.float32)

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        frame_path, label = self.data[idx]
        try:
            image = Image.open(frame_path).convert("RGB")
            if self.transform:
                image = self.transform(image)
        except:
            image = torch.zeros(3, 224, 224)

        return image, torch.tensor(label, dtype=torch.long)

# Transforms with Augmentation
train_transform = transforms.Compose([
    transforms.Resize((256, 256)),
    transforms.RandomCrop(224),
    transforms.RandomHorizontalFlip(),
    transforms.ColorJitter(0.2, 0.2, 0.2),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
])

val_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
])

# Initialize Datasets
dataset_path = "/content/Cleaned_SignLanguageDataset"
train_dataset = SignLanguageDataset(dataset_path, transform=train_transform)
val_dataset = SignLanguageDataset(dataset_path, transform=val_transform)

# Balanced DataLoader
def create_loaders(dataset, batch_size=16, shuffle=True):
    labels = [item[1] for item in dataset.data]
    sampler = torch.utils.data.WeightedRandomSampler(
        weights=dataset.class_weights[labels],
        num_samples=len(dataset),
        replacement=True
    )
    return DataLoader(dataset, batch_size=batch_size, sampler=sampler)

train_loader = create_loaders(train_dataset)
val_loader = DataLoader(val_dataset, batch_size=16, shuffle=False)

# Enhanced Model with Attention
class CNNLSTMAttention(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.cnn = models.mobilenet_v2(weights='IMAGENET1K_V1').features
        self.pool = nn.AdaptiveAvgPool2d((1, 1))

        # LSTM with attention
        self.lstm = nn.LSTM(1280, 512, num_layers=2, bidirectional=True, dropout=0.3)
        self.attention = nn.Sequential(
            nn.Linear(1024, 256),
            nn.Tanh(),
            nn.Linear(256, 1),
            nn.Softmax(dim=1)
        )
        # Classifier
        self.fc = nn.Sequential(
            nn.Dropout(0.5),
            nn.Linear(1024, num_classes))

        # Initialize LSTM
        for name, param in self.lstm.named_parameters():
            if 'weight_ih' in name:
                nn.init.xavier_normal_(param)
            elif 'weight_hh' in name:
                nn.init.orthogonal_(param)
            if 'bias' in name:
                param.data.fill_(0.01)

    def forward(self, x):
        batch_size, seq_len = x.size(0), x.size(1)
        x = x.view(-1, *x.shape[2:])
        x = self.cnn(x)
        x = self.pool(x).view(batch_size, seq_len, -1)

        lstm_out, _ = self.lstm(x)
        attention_w = self.attention(lstm_out)
        context = torch.sum(attention_w * lstm_out, dim=1)

        return self.fc(context)

# Initialize Model
model = CNNLSTMAttention(num_classes=len(train_dataset.classes)).to(device)

# Differential Learning Rates
optimizer = optim.AdamW([
    {'params': model.cnn.parameters(), 'lr': 1e-5},
    {'params': model.lstm.parameters(), 'lr': 1e-4},
    {'params': model.fc.parameters(), 'lr': 1e-4}
], weight_decay=1e-4)

scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, 'max', patience=2, factor=0.5)

# Enhanced Training (No Early Stopping)
def train_model(model, train_loader, val_loader, epochs=50):
    best_acc = 0.0
    criterion = nn.CrossEntropyLoss(weight=train_dataset.class_weights.to(device))

    for epoch in range(epochs):
        model.train()
        train_loss, correct, total = 0.0, 0, 0

        # Training
        for inputs, labels in tqdm(train_loader, desc=f"Epoch {epoch+1}"):
            inputs, labels = inputs.to(device), labels.to(device)

            optimizer.zero_grad()
            outputs = model(inputs.unsqueeze(1))  # Add sequence dimension
            loss = criterion(outputs, labels)
            loss.backward()

            # Gradient clipping
            nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()

            train_loss += loss.item()
            _, predicted = outputs.max(1)
            total += labels.size(0)
            correct += predicted.eq(labels).sum().item()

        # Validation
        model.eval()
        val_loss, val_correct, val_total = 0.0, 0, 0
        with torch.no_grad():
            for inputs, labels in val_loader:
                inputs, labels = inputs.to(device), labels.to(device)
                outputs = model(inputs.unsqueeze(1))
                loss = criterion(outputs, labels)

                val_loss += loss.item()
                _, predicted = outputs.max(1)
                val_total += labels.size(0)
                val_correct += predicted.eq(labels).sum().item()

        # Metrics
        train_acc = correct / total
        val_acc = val_correct / val_total
        scheduler.step(val_acc)

        # Save the best model based on validation accuracy
        if val_acc > best_acc:
            best_acc = val_acc
            torch.save(model.state_dict(), 'best_model.pth')

        print(f"Epoch {epoch+1}:")
        print(f"Train Loss: {train_loss/len(train_loader):.4f} | Acc: {train_acc:.4f}")
        print(f"Val Loss: {val_loss/len(val_loader):.4f} | Acc: {val_acc:.4f}\n")

    model.load_state_dict(torch.load('best_model.pth'))
    return model

# Train the model
model = train_model(model, train_loader, val_loader)

# Inference Function
def predict_video(video_path, model, transform):
    model.eval()
    frames = []

    # Load and process frames
    frame_files = sorted([f for f in os.listdir(video_path) if f.endswith(".jpg")],
                        key=lambda x: int(x.split('_')[-1].split('.')[0]))

    for frame_file in frame_files[:32]:  # Take first 32 frames
        try:
            img = Image.open(os.path.join(video_path, frame_file)).convert("RGB")
            frames.append(transform(img))
        except:
            frames.append(torch.zeros(3, 224, 224))

    # Pad if needed
    if len(frames) < 32:
        frames += [torch.zeros(3, 224, 224)] * (32 - len(frames))

    frames = torch.stack(frames).unsqueeze(0).to(device)  # Add batch dim

    with torch.no_grad():
        outputs = model(frames.unsqueeze(1))  # Add sequence dim
        _, pred = outputs.max(1)

    return train_dataset.classes[pred.item()]

# Example usage:
# prediction = predict_video("path/to/test_video", model, val_transform)
# print("Predicted:", prediction)
```

```python
import cv2
import torch
import numpy as np
from PIL import Image
from torchvision import transforms
import os

# Set device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Define the model architecture
class CNNLSTMAttention(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.cnn = models.mobilenet_v2(weights=None).features
        self.pool = nn.AdaptiveAvgPool2d((1, 1))
        self.lstm = nn.LSTM(1280, 512, num_layers=2, bidirectional=True, dropout=0.3)
        self.attention = nn.Sequential(
            nn.Linear(1024, 256),
            nn.Tanh(),
            nn.Linear(256, 1),
            nn.Softmax(dim=1)
        )
        self.fc = nn.Sequential(
            nn.Dropout(0.5),
            nn.Linear(1024, num_classes))

    def forward(self, x):
        batch_size, seq_len = x.size(0), x.size(1)
        x = x.view(-1, *x.shape[2:])
        x = self.cnn(x)
        x = self.pool(x).view(batch_size, seq_len, -1)
        lstm_out, _ = self.lstm(x)
        attention_w = self.attention(lstm_out)
        context = torch.sum(attention_w * lstm_out, dim=1)
        return self.fc(context)

# Load the trained model
def load_model(model_path, num_classes):
    model = CNNLSTMAttention(num_classes=num_classes)
    model.load_state_dict(torch.load(model_path, map_location=device))
    model.to(device)
    model.eval()
    return model

# Video processing and prediction
def predict_video(video_path, model, transform, class_names):
    # Read video and extract frames
    cap = cv2.VideoCapture(video_path)
    frames = []
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    if total_frames < 1:
        return "Error: Video has no frames"

    # Select 32 evenly spaced frames
    frame_indices = np.linspace(0, total_frames-1, 32, dtype=int)

    for idx in frame_indices:
        cap.set(cv2.CAP_PROP_POS_FRAMES, idx)
        ret, frame = cap.read()
        if ret:
            frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            frame = Image.fromarray(frame)
            frame = transform(frame)
        else:
            frame = torch.zeros(3, 224, 224)
        frames.append(frame)
    cap.release()

    # Convert to tensor
    frames_tensor = torch.stack(frames).unsqueeze(0).to(device)  # Shape: [1, 32, 3, 224, 224]

    # Predict
    with torch.no_grad():
        outputs = model(frames_tensor)
        _, pred = torch.max(outputs, 1)

    return class_names[pred.item()]

# Example usage
if __name__ == "__main__":
    # Load class names (replace with your class names)
    dataset_path = "/content/Cleaned_SignLanguageDataset"
    class_names = sorted(os.listdir(dataset_path))

    # Initialize components
    model = load_model("best_model.pth", num_classes=len(class_names))
    val_transform = transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
    ])

    # Make prediction
    video_path = "/content/are you free today.mp4"  # Replace with your video path
    prediction = predict_video(video_path, model, val_transform, class_names)
    print(f"Predicted sign: {prediction}")
```

```python
import cv2
import torch
import numpy as np
from PIL import Image
import torch.nn as nn
import torchvision.models as models
from torchvision import transforms
from sklearn.utils.class_weight import compute_class_weight
from tqdm import tqdm

# ================== 1. ENHANCED DATA PROCESSING ==================
class SignLanguageDataset(Dataset):
    def __init__(self, root_dir, transform=None, seq_length=32):
        self.root_dir = root_dir
        self.transform = transform
        self.seq_length = seq_length
        self.classes = sorted(os.listdir(root_dir))
        self.label_map = {cls: idx for idx, cls in enumerate(self.classes)}

        # Improved class weighting using sqrt for smoother balance
        labels = []
        self.data = []
        for cls in self.classes:
            cls_path = os.path.join(root_dir, cls)
            videos = [v for v in os.listdir(cls_path) if os.path.isdir(os.path.join(cls_path, v))]
            for video in videos:
                video_path = os.path.join(cls_path, video)
                frames = sorted([f for f in os.listdir(video_path) if f.endswith(".jpg")],
                                key=lambda x: int(x.split('_')[-1].split('.')[0]))
                self.data.append((video_path, self.label_map[cls]))
                labels.append(self.label_map[cls])

        # Smoother class weights
        class_weights = compute_class_weight('balanced', classes=np.unique(labels), y=labels)
        self.class_weights = torch.tensor(class_weights, dtype=torch.float32)

    def __getitem__(self, idx):
        video_path, label = self.data[idx]
        frames = self._load_frames(video_path)
        return frames, torch.tensor(label, dtype=torch.long)

    def _load_frames(self, video_path):
        frame_files = sorted(os.listdir(video_path))
        num_frames = len(frame_files)

        # Dynamic frame selection with temporal awareness
        selected_frames = []
        if num_frames >= self.seq_length:
            indices = np.linspace(0, num_frames-1, self.seq_length, dtype=int)
        else:
            indices = np.arange(num_frames)
            indices = np.append(indices, [num_frames-1]*(self.seq_length-num_frames))

        for idx in indices:
            frame_path = os.path.join(video_path, frame_files[idx])
            try:
                img = Image.open(frame_path).convert("RGB")
                if self.transform:
                    img = self.transform(img)
            except:
                img = torch.zeros(3, 224, 224)
            selected_frames.append(img)

        return torch.stack(selected_frames)

# ================== 2. IMPROVED MODEL ARCHITECTURE ==================


# ================== ORIGINAL MODEL ARCHITECTURE ==================
class CNNLSTMAttention(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.cnn = models.mobilenet_v2(weights=None).features
        self.pool = nn.AdaptiveAvgPool2d((1, 1))

        self.lstm = nn.LSTM(1280, 512, num_layers=2, bidirectional=True, dropout=0.3)
        self.attention = nn.Sequential(
            nn.Linear(1024, 256),
            nn.Tanh(),
            nn.Linear(256, 1),
            nn.Softmax(dim=1))

        self.fc = nn.Sequential(
            nn.Dropout(0.5),
            nn.Linear(1024, num_classes))

    def forward(self, x):
        batch_size, seq_len = x.size(0), x.size(1)
        x = x.view(-1, *x.shape[2:])
        x = self.cnn(x)
        x = self.pool(x).view(batch_size, seq_len, -1)

        lstm_out, _ = self.lstm(x)
        attention_w = self.attention(lstm_out)
        context = torch.sum(attention_w * lstm_out, dim=1)

        return self.fc(context)

# ================== INFERENCE PIPELINE ==================
class SignLanguagePredictor:
    def __init__(self, model_path, dataset_path):
        self.device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
        self.class_names = sorted(os.listdir(dataset_path))
        self.model = self._load_model(model_path)
        self.transform = transforms.Compose([
            transforms.Resize((224, 224)),
            transforms.ToTensor(),
            transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
        ])

    def _load_model(self, model_path):
        model = CNNLSTMAttention(num_classes=len(self.class_names))
        model.load_state_dict(torch.load(model_path, map_location=self.device))
        model.to(self.device)
        model.eval()
        return model

    def _process_video(self, video_path):
        cap = cv2.VideoCapture(video_path)
        frames = []
        total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

        # Extract 32 evenly spaced frames
        frame_indices = np.linspace(0, total_frames-1, 32, dtype=int)

        for idx in frame_indices:
            cap.set(cv2.CAP_PROP_POS_FRAMES, idx)
            ret, frame = cap.read()
            if ret:
                frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
                frame = Image.fromarray(frame)
                frame = self.transform(frame)
            else:
                frame = torch.zeros(3, 224, 224)
            frames.append(frame)

        cap.release()
        return torch.stack(frames).unsqueeze(0).to(self.device)  # Add batch dim

    def predict(self, video_path):
        try:
            inputs = self._process_video(video_path)
            with torch.no_grad():
                outputs = self.model(inputs)
                _, pred = torch.max(outputs, 1)

            return self.class_names[pred.item()]
        except Exception as e:
            return f"Error: {str(e)}"

# ================== USAGE ==================
if __name__ == "__main__":
    # Initialize predictor
    dataset_path = "/content/Cleaned_SignLanguageDataset"  # Same as training data
    predictor = SignLanguagePredictor("best_model.pth", dataset_path)

    # Make prediction
    video_path = "/content/you can do it.mp4"  # Replace with your video
    prediction = predictor.predict(video_path)
    print("Predicted Sign:", prediction)
```

```python
import os
import cv2
import torch
import numpy as np
from PIL import Image
import torch.nn as nn
import torchvision.models as models
from torchvision import transforms

class SignLanguagePredictor:
    def __init__(self, model_path, dataset_path, confidence_threshold=0.5):
        self.device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
        self.class_names = sorted(os.listdir(dataset_path))
        self.model = self._load_model(model_path)
        self.confidence_threshold = confidence_threshold
        self.transform = transforms.Compose([
            transforms.Resize((224, 224)),
            transforms.ToTensor(),
            transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
        ])

    def _load_model(self, model_path):
        model = CNNLSTMAttention(num_classes=len(self.class_names))
        model.load_state_dict(torch.load(model_path, map_location=self.device))
        model.to(self.device)
        model.eval()
        return model

    def _process_video(self, video_path):
        cap = cv2.VideoCapture(video_path)
        frames = []
        total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

        frame_indices = np.linspace(0, total_frames-1, 32, dtype=int)

        for idx in frame_indices:
            cap.set(cv2.CAP_PROP_POS_FRAMES, idx)
            ret, frame = cap.read()
            if ret:
                frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
                frame = Image.fromarray(frame)
                frame = self.transform(frame)
            else:
                frame = torch.zeros(3, 224, 224)
            frames.append(frame)

        cap.release()
        return torch.stack(frames).unsqueeze(0).to(self.device)

    def predict(self, video_path):
        try:
            # Get base filename without extension
            filename = os.path.splitext(os.path.basename(video_path))[0]

            inputs = self._process_video(video_path)

            with torch.no_grad():
                outputs = self.model(inputs)
                probabilities = torch.nn.functional.softmax(outputs, dim=1)
                conf, pred = torch.max(probabilities, 1)
                confidence = conf.item()

            predicted_class = self.class_names[pred.item()]

            if confidence < self.confidence_threshold:
                return filename

            return predicted_class

        except Exception as e:
            return filename

# Usage
if __name__ == "__main__":
    dataset_path = "/content/Cleaned_SignLanguageDataset"
    predictor = SignLanguagePredictor("best_model.pth", dataset_path, confidence_threshold=0.8)

    video_path = "/content/can i help you.mp4"
    prediction = predictor.predict(video_path)
    print("Final Output:", prediction)
```
